# String indexing

```python
s = 'prakash'
print(len(s))
print(s[-1])   # 'h'
print(s[2:])   # 'akash'
print(s[:3])     # 'pra'
print(s[3:])     # 'kash'
print(s[:])      # 'prakash'
print(s[0:6:2])  # 'paa'
print(s[::-1])   # 'hsakarp'
print(s[-7:-2])   # 'praka
print(s[2:-2])   # 'aka'
```
