```python
inventory = {"laptop":15, "Mouse":5, "keyboard":4, "Monitor":25,"Earphones":2}

# Identify the products that are running "out of stock"(0), "running low"(<=5), "sufficient stock"(>5) ;

# which products are in each categoty and how many products do we need to reach sufficient stock.
for j in inventory.items():
    product,stock = j[0],j[1]
    if (stock <= 0):
        catagory="Out of stock"
        print(f"ProductName:{product}\nCatagry:{catagory}")
    elif(stock <= 5):
        catagory = "Running low stock"
        print(f"ProductName:{product}\nCatagry:{catagory}")
    else:
        catagory = "Sufficient stock"
        print(f"ProductName:{product}\nCatagry:{catagory}")


```

```python
# Based on employee rating, decide their salary increment

# rating 4 or 5 -> 20% increment

# rating 3-> 10%

# rating 1 or 2 -> no increment

ratings = {"RAVI": 5, "ANIL": 3,"Vijay": 2, "Vinay":2}
salaries = {"RAVI": 25000, "ANIL": 45000,"Vijay": 75000, "Vinay":55000}

# display following in the output:
# name, rating, old salary, increment, new salary
updated_salaries = {}
for i in ratings.items():
    name, rating = i[0], i[1]
    salary = salaries[name]
    # print(name, rating, salary)
    if (rating == 3):
        anual_income = salary * 12
        increment = anual_income * 0.20
        total_salary = increment + anual_income
        updated_salaries[name] = total_salary
        print(f"Name:{name}\nIncrement:{increment}\nTotalSalary:{total_salary}") 
        print("=" * 30)
    elif(rating >= 4):
        anual_income = salary * 12
        increment = anual_income * 0.20
        total_salary = increment + anual_income
        updated_salaries[name] = total_salary
        print(f"Name:{name}\nIncrement:{increment}\nTotalSalary:{total_salary}")
        print("=" * 30)

print(updated_salaries)  
```
