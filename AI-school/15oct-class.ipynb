```python
cart = [
    {
        "product": "Laptop",
        "Category": 'Electronics',
        "details": (55000,1)
    },
    {
        "product": "Mouse",
        "Category": 'Electronics',
        "details": (800,2)
    },
    {
        "product": "Keyboard",
        "Category": 'Electronics',
        "details": (2500,1)
    },
    {
        "product": "Monitor",
        "Category": 'Electronics',
        "details": (12000,1)
    },
    {
        "product": "Shoes",
        "Category": "Accessories",
        "details": (2500,4)
    }    
]
# Print each product name, price and quantity
```

```python
for prod in cart:
    # print(prod)
    prod_name = prod["product"]
    category = prod["Category"]
    price = prod["details"][0]
    qty = prod["details"][1]
    print("Product Name:",prod["product"])
    print("Catogery:",prod["Category"])
    print("Acutal Price:", prod["details"][0])
    print("Quanitiy:",prod["details"][1])
    print("Totla cost",price * qty )

    if category == "Accessories":
        # Apply 5% off in the price
        off_price = price * 0.05
        dicount_price = price - off_price
        print("5% off:", off_price)
        print("Price After discount:", dicount_price)
        print("Total cost after discount:", dicount_price * qty)
    elif category == "Electronics":
        # Apply 10% off in the price
        off_price = price * 0.10
        dicount_price = price - off_price
        print("10% off:", off_price)
        print("Price After discount:", dicount_price)
        print("Total cost after discount:", dicount_price * qty)        
    print("="*20)

```
