```python
class math_operations:
    a = 100
    def addition(self):
        n = int(input("Enter a number: "))
        return a + n

    def multiplication(self):
        n = int(input("Enter a number: "))
        return a * n
```

```python
output = math_operations()
# output.a
# output.addition()
print(output.a)
print(output.addition())
print(output.multiplication())
```
