```python
import pyfiglet
from rich import print, box
from rich.panel import Panel
from rich.align import Align

text = "\n".join(l for l in pyfiglet.figlet_format("Part-1", font="moscow").splitlines() if l.strip())
print(Panel(Align.center(f"[yellow]{text}[/yellow]"),box=box.DOUBLE_EDGE, border_style="yellow", width=50, padding=(1, 2)))
```
